### Hàm kiểm tra số nguyên tố

```python
## ham kiem tra so nguyen to
def ktsnt(n):
    if n < 2: # 0, 1 khong la so nguyen to
        return False
    for i in range(2, int(n**(0,5))+1): # can bac hai cua n
        if (n % 1 == 0):
            return False
    return True
### nhap vao so nguyen can kiem tra
a = int(input("Nhap vao so nguyen:"))
if ktsnt(a):
    print("%d la so nguyen to" % a)
else:
    print("%d khong la so nguyen to" % a)
```

### Hàm liệt kê các số nguyên tố

```python
### ham liet ke cac so nguyen to nho hon n 
def lietkesnt(N):
    if N <= 2:
        print("Không có số nguyên tố nào nhỏ hơn", N)
    else:
        print(f"Các số nguyên tố nhỏ hơn {N}:")
        for i in range(2, N):
            if ktsnt(i):
                print(i, end=" ")

N = int(input("Nhập N: "))
lietkesnt(N)
```

### Hàm in ra số nguyên tố đầu tiên

```python
import math

def ktsnt(n):
    """Kiem tra so nguyen to"""
    if n < 2:
        return False
    if n % 2 == 0:
        return n == 2
    r = int(math.isqrt(n))
    for i in range(3, r + 1, 2):
        if n % i == 0:
            return False
    return True

### ham Liet ke M so nguyen to dau tien
def lietkeMsnt(M):
    bien_dem = 0      # dem: tu 0 den M-1
    t = 2             # 0 va 1 khong la so nguyen to
    while bien_dem < M:
        if ktsnt(t):
            print("%d " % t, end="")
            bien_dem += 1
        t += 1

# Vi du su dung
M = 10
lietkeMsnt(M)
```

```python
def lietke2(M,N):
    q = -1
    for i in range(M,N+1):
        if (i%9==0) and (i%7==0):
            q = i
            break
    return q
M = 64
N = 300
q = lietke2(M,N)
if (q == -1):
    print("khong co so thoa dieu kien")
else:
    print("%d " %q)
```

### Hàm đếm số lượng nguyên âm (a e i o u), phụ âm có trong chuỗi

```python
# Chuỗi test
test_string = "  !! fdafdaf "

# Các ký tự đặc biệt loại bỏ
valid_symbols = "!@#$%^&*()_+-=[]{}/<>?,."

def dem_nguyenam_phuam(string):
    nguyenam = 0
    phuam = 0
    for char in string:
        # bỏ qua khoảng trắng và ký hiệu đặc biệt
        if char != " " and char not in valid_symbols:
            if char in "aeiouAEIOU":   # kiểm tra nguyên âm
                nguyenam += 1
            else:                      # còn lại là phụ âm
                phuam += 1
    return nguyenam, phuam

# Test
print("Độ dài chuỗi:", len(test_string))
v, c = dem_nguyenam_phuam(test_string)
print("Nguyên âm:", v, " | Phụ âm:", c)
```

```python
# Các ký tự đặc biệt cần loại bỏ
valid_symbols = "!@#$%^&*()_+-={}[]|\\:;\"'<>,.?/"

def dem_nguyenam_phuam_2(string):
    nguyenam = 0
    phuam = 0
    vowels = "aeiouAEIOU"   # tập nguyên âm
    
    for char in string:
        if char != ' ' and char not in valid_symbols:  # bỏ qua khoảng trắng và ký hiệu
            if char in vowels:   # nếu là nguyên âm
                nguyenam += 1
            else:                # còn lại là phụ âm
                phuam += 1
    return nguyenam, phuam


# Test
test_string = " !! fdafdaf "
v, c = dem_nguyenam_phuam_2(test_string)
print("Nguyên âm:", v, "| Phụ âm:", c)
```

# Cấu trúc lặp

## Bài 1: Kiểm tra số nguyên dương n

```python
n = int(input("Nhập số nguyên dương n (>0): "))

# 1.1 Kiểm tra số đối xứng (palindrome)
if str(n) == str(n)[::-1]:
    print(f"{n} là số đối xứng")
else:
    print(f"{n} không phải số đối xứng")

# 1.2 Kiểm tra số chính phương
if int(math.sqrt(n))**2 == n:
    print(f"{n} là số chính phương")
else:
    print(f"{n} không phải số chính phương")
    
# 1.3 Kiểm tra số nguyên tố
def is_prime(x):
    if x < 2: return False
    for i in range(2, int(math.sqrt(x)) + 1):
        if x % i == 0:
            return False
    return True

print(f"{n} là số nguyên tố" if is_prime(n) else f"{n} không phải số nguyên tố")

# 1.4 Chữ số lớn nhất và nhỏ nhất
digits = [int(d) for d in str(n)]
print("Chữ số lớn nhất:", max(digits))
print("Chữ số nhỏ nhất:", min(digits))

# 1.5 Các chữ số tăng dần / giảm dần
if digits == sorted(digits):
    print("Các chữ số tăng dần")
elif digits == sorted(digits, reverse=True):
    print("Các chữ số giảm dần")
else:
    print("Các chữ số không tăng dần cũng không giảm dần")
```

## Bài 2: Tính toán với số nguyên dương n

```python
n = int(input("\nNhập số nguyên dương n: "))

# a) S = 1 + 2 + ... + n
S1 = sum(range(1, n+1))
print("S1 =", S1)

# b) S = 1^2 + 2^2 + ... + n^2
S2 = sum(i*i for i in range(1, n+1))
print("S2 =", S2)

# c) S = 1 + 1/2 + ... + 1/n
S3 = sum(1/i for i in range(1, n+1))
print("S3 =", S3)

# d) S = 1*2*...*n = n!
S4 = math.factorial(n)
print("S4 (n!) =", S4)

# e) S = 1! + 2! + ... + n!
S5 = sum(math.factorial(i) for i in range(1, n+1))
print("S5 =", S5)
```
