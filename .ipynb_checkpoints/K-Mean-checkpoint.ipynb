# Bài thực hành 1

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

plt.style.use('seaborn-v0_8')
sns.set_theme(style="whitegrid", palette="muted")

# Đọc dữ liệu chim cánh cụt
penguins = pd.read_csv("C:/Users/Admin/Downloads/penguins_size.csv")

print("Kích thước dữ liệu:", penguins.shape)
display(penguins.head())
```

```python
# Thông tin chung
penguins.info()

# Kiểm tra giá trị thiếu
print("\nSố lượng giá trị thiếu:")
print(penguins.isnull().sum())

# Xem phân bố dữ liệu theo loài
print("\nPhân bố các loài chim:")
print(penguins['Species'].value_counts())
```

```python
# Loại bỏ dòng có giá trị thiếu
penguins_clean = penguins.dropna()

# Giữ lại chỉ các cột số (K-Means không xử lý được chữ)
penguins_num = penguins_clean.select_dtypes(include=['number'])

print("Các cột được giữ lại cho mô hình:", penguins_num.columns.tolist())
print("Kích thước dữ liệu sau khi làm sạch:", penguins_num.shape)
```

```python
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(penguins_num)

print("Kích thước dữ liệu sau chuẩn hóa:", X_scaled.shape)
```

```python
inertia = []
K = range(2, 10)

for k in K:
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(X_scaled)
    inertia.append(km.inertia_)

plt.figure(figsize=(8,5))
plt.plot(K, inertia, 'bo-', markersize=6)
plt.title('Phương pháp Elbow - Chọn số cụm tối ưu', fontsize=14, weight='bold')
plt.xlabel('Số cụm (k)')
plt.ylabel('Inertia (SSE)')
plt.show()
```

```python
for k in range(2, 10):
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(X_scaled)
    score = silhouette_score(X_scaled, km.labels_)
    print(f"K = {k}, Silhouette Score = {score:.3f}")
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X_scaled)

# Gán nhãn cụm vào dataset đã làm sạch
penguins_clean['Cluster'] = kmeans.labels_

# Kiểm tra số dòng khớp
print("Số dòng dữ liệu:", len(penguins_clean))
print("Số nhãn cụm:", len(kmeans.labels_))
```

```python
# Số lượng phần tử mỗi cụm
print("\nSố lượng chim trong mỗi cụm:")
print(penguins_clean['Cluster'].value_counts())
# Trung bình các đặc trưng trong từng cụm
cluster_means = penguins_clean.groupby('Cluster')[['Culmen Length (mm)', 'Culmen Depth (mm)', 
                                                   'Flipper Length (mm)', 'Body Mass (g)']].mean().round(2)
display(cluster_means)
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    x='Culmen Length (mm)',
    y='Culmen Depth (mm)',
    hue='Cluster',
    data= penguins_clean,
    palette='Set2',
    s=80
)
plt.title('Kết quả phân cụm K-Means theo đặc trưng mỏ chim', fontsize=14, weight='bold')
plt.xlabel('Culmen Length (mm)')
plt.ylabel('Culmen Depth (mm)')
plt.legend(title='Cụm')
plt.tight_layout()
plt.show()
```

```python
comparison = pd.crosstab(penguins_clean['Species'], penguins_clean['Cluster'])
print("So sánh cụm với loài thật:")
display(comparison)
```

```python
import plotly.express as px

fig = px.scatter_3d(
    penguins_clean,
    x='Culmen Length (mm)',
    y='Culmen Depth (mm)',
    z='Flipper Length (mm)',
    color='Cluster',
    title='Phân cụm chim cánh cụt bằng K-Means (3D)',
    color_continuous_scale='Viridis'
)
fig.show()
```

# Bài thực hành 2

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

plt.style.use('seaborn-v0_8')
sns.set_theme(style='whitegrid', palette='muted')

# Đọc file Excel
data = pd.read_excel("C:/Users/Admin/Downloads/Online Retail.xlsx/Online Retail.xlsx")

print("Kích thước dữ liệu:", data.shape)
display(data.head())
```

```python
data.info()

# Kiểm tra dữ liệu thiếu
print("\nSố lượng giá trị thiếu:")
print(data.isnull().sum())

# Xem mô tả thống kê
display(data.describe())

# Số quốc gia & khách hàng
print("\nSố khách hàng khác nhau:", data['CustomerID'].nunique())
print("Số quốc gia khác nhau:", data['Country'].nunique())
```

```python
# Loại bỏ các dòng thiếu CustomerID (vì không biết ai mua)
data = data.dropna(subset=['CustomerID'])

# Loại bỏ đơn hàng hủy (InvoiceNo bắt đầu bằng 'C')
data = data[~data['InvoiceNo'].astype(str).str.startswith('C')]

# Loại bỏ giá trị âm hoặc 0
data = data[(data['Quantity'] > 0) & (data['UnitPrice'] > 0)]

# Tạo cột doanh thu
data['TotalPrice'] = data['Quantity'] * data['UnitPrice']

print("Kích thước dữ liệu sau khi làm sạch:", data.shape)
display(data.head())
```

```python
# Chọn ngày cuối cùng trong dữ liệu làm mốc tính Recency
snapshot_date = data['InvoiceDate'].max()

# Tính toán các đặc trưng
rfm = data.groupby('CustomerID').agg({
    'InvoiceDate': lambda x: (snapshot_date - x.max()).days,  # Recency
    'InvoiceNo': 'count',                                     # Frequency
    'TotalPrice': 'sum'                                       # Monetary
}).reset_index()

rfm.rename(columns={'InvoiceDate':'Recency', 'InvoiceNo':'Frequency', 'TotalPrice':'Monetary'}, inplace=True)

display(rfm.head())
```

```python
scaler = MinMaxScaler()
rfm_scaled = scaler.fit_transform(rfm[['Recency','Frequency','Monetary']])
```

```python
inertia = []
K = range(2, 10)

for k in K:
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(rfm_scaled)
    inertia.append(km.inertia_)

plt.figure(figsize=(8,5))
plt.plot(K, inertia, 'bo-')
plt.title('Phương pháp Elbow - Chọn số cụm tối ưu', fontsize=14, weight='bold')
plt.xlabel('Số cụm (k)')
plt.ylabel('Inertia (SSE)')
plt.show()
```

```python
for k in range(2, 10):
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(rfm_scaled)
    score = silhouette_score(rfm_scaled, km.labels_)
    print(f"K = {k}, Silhouette Score = {score:.3f}")
```

```python
kmeans = KMeans(n_clusters=4, random_state=42)
kmeans.fit(rfm_scaled)

rfm['Cluster'] = kmeans.labels_
rfm.head()
```

```python
# Trung bình các đặc trưng theo cụm
cluster_summary = rfm.groupby('Cluster')[['Recency','Frequency','Monetary']].mean().round(1)
display(cluster_summary)

# Số lượng khách hàng mỗi cụm
print("\nSố lượng khách hàng trong mỗi cụm:")
print(rfm['Cluster'].value_counts())
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    x='Frequency',
    y='Monetary',
    hue='Cluster',
    data=rfm,
    palette='Set2',
    s=80
)
plt.title('Phân cụm khách hàng theo Frequency và Monetary', fontsize=14, weight='bold')
plt.xlabel('Frequency (Số hóa đơn)')
plt.ylabel('Monetary (Tổng chi tiêu)')
plt.legend(title='Cụm')
plt.tight_layout()
plt.show()
```
