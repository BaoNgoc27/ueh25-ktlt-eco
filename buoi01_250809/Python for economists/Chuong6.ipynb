## 6.2 if-else Conditional Instructions

```python
temperature = 26
heat = ""
 
if temperature > 28:
  heat = "hot"
else:
  heat = "cold"
 
print("It is " + heat + " out there")
```

## 6.3 if-elif Conditional Instructions

```python
temperature = -4
heat = ""
 
if temperature > 28:
  heat = "hot"
elif temperature <= 28 and temperature > 15:
  heat = "not too hot"
elif temperature <= 15 and temperature > 0:
  heat = "cold"
else:
 heat = "very cold"
 
print("It is " + heat + " out there")
```
