### Tính số ngày giữa hai mốc thời gian

beginning = date(2021, 2, 28)
end = date(2021, 3, 1)
duration = end - beginning
duration
datetime.timedelta(1)

- Khi trừ hai date, Python trả về một datetime.timedelta biểu thị khoảng cách giữa chúng.

- Năm 2021 không nhuận nên từ 28/2 đến 1/3 chỉ cách 1 ngày → datetime.timedelta(1).

### Trường hợp năm nhuận

beginning_leap = date(2020, 2,28)
end_leap = date(2020, 3, 1)
beginning_leap = end_leap- beginning_leap
beginning_leap
datetime.timedelta(2)

- Năm 2020 là năm nhuận, nên sau 28/2 là 29/2, rồi mới tới 1/3.

- Khoảng cách từ 28/2 đến 1/3 là 2 ngày → datetime.timedelta(2).

### Cộng thêm thời gian vào datetime

```python
debut = datetime(2018, 12, 31, 23, 59, 59)
print(debut + timedelta(seconds = 1))
```
